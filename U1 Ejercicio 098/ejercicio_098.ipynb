## Instrucciones
Implementa la clase `MascotaClinicaV10` respetando los siguientes requerimientos:

1. **`__init__(self, nombre: str, edad: float)`**:
   - Inicializa las variables de instancia: `nombre` (str), `edad` (float), `historial_vacunas` (list vacía) y `activo` (bool = True).

2. **`evaluar_condicion(self, limite: float, factor: float) -> float`**:
   - Usa operadores aritméticos, lógicos y condicionales (`if / elif / else`).
   - Si `edad` >= limite y `activo` es True: retorna `edad` * factor.
   - Elif `edad` < limite y `edad` > 0: retorna `edad` + 10.0.
   - Else: retorna 0.0.

3. **`procesar_registros(self, datos: list) -> float`**:
   - Usa un ciclo `for` para recorrer `datos` (lista de floats/ints/None).
   - Ignora los elementos que sean `None`.
   - Acumula los valores válidos en `edad` y agrega cada valor a `historial_vacunas`.
   - Si `edad` supera los 500.0, desactiva la entidad (`activo = False`).
   - Retorna el valor final de `edad`.

4. **`acumular_hasta_objetivo(self, objetivo: float, paso: float) -> int`**:
   - Usa un ciclo `while` para incrementar `edad` en `paso` mientras `edad` sea estrictamente menor que `objetivo`.
   - Retorna la cantidad de iteraciones/pasos realizados.


```python
# Escribe tu solución aquí
class MascotaClinicaV10:
    def _init_(self, nombre: str, edad: float):
        self.nombre = nombre
        self.edad = edad
        self.historial_vacunas = []
        self.activo = True

    def evaluar_condicion(self, limite: float, factor: float) -> float:
        if self.edad >= limite and self.activo:
            return self.edad * factor
        elif self.edad < limite and self.edad > 0:
            return self.edad + 10.0
        else:
            return 0.0

    def procesar_registros(self, datos: list) -> float:
        for valor in datos:
            if valor is not None:
                self.edad += valor
                self.historial_vacunas.append(valor)

                if self.edad > 500.0:
                    self.activo = False

        return self.edad

    def acumular_hasta_objetivo(self, objetivo: float, paso: float) -> int:
        contador = 0

        while self.edad < objetivo:
            self.edad += paso
            contador += 1

        return contador
```
